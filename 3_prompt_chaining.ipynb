```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
from langchain_huggingface import ChatHuggingFace,HuggingFaceEndpoint
from dotenv import load_dotenv
```

```python
load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V4.1-Flash",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)
```

```python
# Create State

class LLMState(TypedDict):

    topic : str
    outline : str
    content : str
```

```python
def topic_outline(state : LLMState) -> LLMState:

    topic = state['topic']

    prompt = f'generate a detailed outline for a blog on the topic  {topic}'

    outline = model.invoke(prompt).content

    state['outline'] = outline

    return state
```

```python
def topic_content(state : LLMState) -> LLMState:

    title = state['topic']
    outline = state['outline']

    prompt = f'Write a detaild block on the title {title} using  the following outline {outline}'

    blog = model.invoke(prompt).content

    state['content'] = blog

    return state

```

```python
# define graph

graph = StateGraph(LLMState)

# nodes
graph.add_node('topic_outline',topic_outline)
graph.add_node('topic_content',topic_content)

# Edges
graph.add_edge(START,'topic_outline')
graph.add_edge('topic_outline','topic_content')
graph.add_edge('topic_content',END)
```

```python
workflow = graph.compile()
```

```python
initial_state = {'topic' : 'Virat kohali in ODI '}
final_state = workflow.invoke(initial_state)
```
